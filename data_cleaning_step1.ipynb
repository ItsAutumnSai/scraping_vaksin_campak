```python
import pandas as pd
import numpy as np

# init
keywords = ['campak', 'vaksin campak', 'vaksin rubella', 'imunisasi',
            'anti-vaksin', 'measless rubella', 'rubella', 'vaksin MR',
            'efek samping vaksin', 'vaksinasi anak', 'anak']
keyword_pattern = '|'.join(keywords)
pd.set_option('display.max_colwidth', None)
pd_threads = pd.read_csv("threads_vaccine.csv", na_values=['NaN', '', 'unknown', -1])
pd_xcom = pd.read_csv("x_vaccine.csv", na_values=['NaN', '', 'unknown', -1])
pd_ig = pd.read_csv("instagram_vaccine.csv", na_values=['NaN', '', 'unknown', -1])
```

```python
pd_threads['postdate'] = pd.to_datetime(pd_threads['postdate'])
pd_threads = pd_threads.drop_duplicates(subset=['tweet'])
test = pd_threads
test = test.drop(columns=['index'])

test = test[test['tweet'].str.contains(keyword_pattern, case=False, na=False)]
test = test.reset_index(drop=True)
print(len(test), 'out of', len(pd_threads), '=', round(len(test)/len(pd_threads), 3))
pd_threads_clean = test
```

```python
pd_xcom['postdate'] = pd.to_datetime(pd_xcom['postdate'])
pd_xcom = pd_xcom.drop_duplicates(subset=['tweet'])
test = pd_xcom
test = test.drop(columns=['index'])

test = test[test['tweet'].str.contains(keyword_pattern, case=False, na=False)]
test = test.reset_index(drop=True)
print(len(test), 'out of', len(pd_xcom), '=', round(len(test)/len(pd_xcom), 3))
pd_xcom_clean = test
```

```python
pd_ig['postdate'] = pd.to_datetime(pd_ig['postdate'])
pd_ig = pd_ig.drop_duplicates(subset=['tweet'])
test = pd_ig
test = test.drop(columns=['index'])

test = test[test['tweet'].str.contains(keyword_pattern, case=False, na=False)]
test['tweet'] = test['tweet'].str.replace(r'\n ', '', regex=True)
test = test.reset_index(drop=True)
print(len(test), 'out of', len(pd_ig), '=', round(len(test)/len(pd_ig), 3))
pd_ig_clean = test
```

```python
pd_fb1 = pd.read_csv("facebook/fb_vaccine_results.csv", na_values=['NaN', '', 'unknown', -1])
pd_fb2 = pd.read_csv("facebook/fb_vaccine_results_old2.csv", na_values=['NaN', '', 'unknown', -1])
pd_fb3 = pd.read_csv("facebook/fb_vaccine_results_old3.csv", na_values=['NaN', '', 'unknown', -1])
pd_fb4 = pd.read_csv("facebook/fb_vaccine_results_old4.csv", na_values=['NaN', '', 'unknown', -1])
pd_fb5 = pd.read_csv("facebook/fb_vaccine_results_old5.csv", na_values=['NaN', '', 'unknown', -1])

pd_fb = pd.concat([pd_fb1, pd_fb2, pd_fb3, pd_fb4, pd_fb5], axis=0, ignore_index = True)

pd_fb['postdate'] = np.nan
pd_fb = pd_fb.drop_duplicates(subset=['tweet'])
test = pd_fb
test = test.drop(columns=['index'])

test = test[test['tweet'].str.contains(keyword_pattern, case=False, na=False)]
test['tweet'] = test['tweet'].str.replace(r'\n ', '', regex=True)
test = test.reset_index(drop=True)
print(len(test), 'out of', len(pd_fb), '=', round(len(test)/len(pd_fb), 3))
pd_fb_clean = test
```

```python
pd_threads_clean['platform'] = 'threads'
pd_xcom_clean['platform'] = 'x'
pd_ig_clean['platform'] = 'instagram'
pd_fb_clean['platform'] = 'facebook'
merged_data = pd.concat([pd_threads_clean, pd_xcom_clean, pd_ig_clean, pd_fb_clean], axis=0, ignore_index=True)
merged_data.to_csv('cleaned_data.csv', index=False)
```

```python
merged_data = pd.read_csv("cleaned_data.csv")
merged_data['postdate'] = pd.to_datetime(merged_data['postdate'])
merged_data_2025 = merged_data[(merged_data['postdate'].dt.year == 2025) | (merged_data['postdate'].isna())]
merged_data_2025.to_csv('cleaned_data_2025.csv', index=False)
```

```python
monthly_counts = merged_data_2025['postdate'].dt.to_period('M').value_counts(dropna=False).sort_index()
platform_counts = merged_data_2025['platform'].value_counts().sort_index()

print(monthly_counts)
print(platform_counts)
print("total rows of data: ", len(merged_data_2025))
```
