```python
import pandas as pd
import numpy as np

#facebook
pd.set_option('display.max_colwidth', None)
data = pd.read_csv('scraping/out/facebook_vaccine_2024_2026.classified.csv', na_values=['NaN', -1])
data = data.dropna(subset=['postdate'])
data['postdate'] = pd.to_datetime(data['postdate'])
data = data.drop_duplicates(subset=['tweet'])
data = data[(data['is_relevant'] == True) | (data["replysource"].notna())]
data = data[data['is_indonesian'] == True]
data = data.rename(columns={'retweets': 'shares'})
data = data.drop(columns=['index', 'views', 'veto_reason', 'has_topic_keyword', 'borderline', 'relevance_score', 'is_indonesian', 'is_relevant'])
# data = data[data['replysource'] == 'https://web.facebook.com/cintaofficial1212/posts/pfbid0eGCKu3qjJ9TBhE9d7NXSAZxUH5bTn1KeTcDjsgACNFYBAqyFTKzvhzyznGddGCJ4l']
# data[data.duplicated(subset=['tweet'], keep=False)]
# data = data[data['retweets'] != 0]
# data = data[data['replysource'].notna()]
print('facebook data count: ', len(data))
data_fb = data
data.head(1)
```

```python
#instagram
data = pd.read_csv('scraping/out/instagram_vaccine_2024_2026.classified.csv', na_values=['NaN', -1])
data_old = pd.read_csv('old_result/instagram_vaccine.csv', na_values=['NaN', -1])
data = pd.concat([data, data_old], axis=0, ignore_index = True)
data = data.dropna(subset=['postdate'])
# data['reposts'] = np.nan
# data = data.rename(columns={'retweets': 'reposts'})
data = data[(data['is_relevant'] == True)]
data = data[data['is_indonesian'] == True]
data['postdate'] = pd.to_datetime(data['postdate'])
data = data.drop_duplicates(subset=['tweet'])
data = data.drop_duplicates(subset=['link'])
data = data.dropna(subset=['postdate'])
data = data.drop(columns=['index', 'retweets', 'views', 'replysource', 'veto_reason', 'has_topic_keyword', 'borderline', 'relevance_score', 'is_indonesian', 'is_relevant'])
# data = data[data['is_relevant'] == True]
# data = data[data['replysource'].notna()]
print('instagram data count: ', len(data))
data_ig = data
data.head(1)
```

```python
#threads
data = pd.read_csv('scraping/out/threads_vaccine_2024_2026.classified.csv', na_values=['NaN', -1])
data_old = pd.read_csv('old_result/threads_vaccine.csv', na_values=['NaN', -1])
data = pd.concat([data, data_old], axis=0, ignore_index = True)
data = data.dropna(subset=['postdate'])
data = data[(data['is_relevant'] == True) | (data["replysource"].notna())]
data = data[data['is_indonesian'] == True]
data['postdate'] = pd.to_datetime(data['postdate'])
data = data.drop_duplicates(subset=['tweet'])
# data = data.drop_duplicates(subset=['link'])
data = data.dropna(subset=['postdate'])
data['tweet'] = data.apply(lambda row: row['tweet'].removeprefix(row['account']), axis=1)
data['tweet'] = data.apply(lambda row: row['tweet'].removeprefix('Verified '), axis=1)
data = data.drop(columns=['index', 'veto_reason', 'has_topic_keyword', 'borderline', 'relevance_score', 'is_indonesian', 'is_relevant'])
# data = data[data['retweets'] != 0]
# data = data[data['replysource'].notna()]
print('threads data count: ', len(data))
data_th = data
data.head(1)
```

```python
#X
data = pd.read_csv('scraping/out/x_vaccine_2024_2026.classified.csv', na_values=['NaN', -1])
data_old = pd.read_csv('old_result/x_vaccine.csv', na_values=['NaN', -1])
data = pd.concat([data, data_old], axis=0, ignore_index = True)
data = data.dropna(subset=['postdate'])
data = data[(data['is_relevant'] == True)]
data = data[data['is_indonesian'] == True]
data['postdate'] = pd.to_datetime(data['postdate'])
data = data.drop_duplicates(subset=['tweet'])
data = data.drop_duplicates(subset=['link'])
data = data.dropna(subset=['postdate'])
data = data[data['index'] != 5]
data = data.drop(columns=['index', 'veto_reason', 'has_topic_keyword', 'borderline', 'relevance_score', 'is_indonesian', 'is_relevant'])
# data = data[data['retweets'] != 0]
# data = data[data['replysource'].notna()]
print('X data count: ', len(data))
data_x = data
data.head(1)
```

```python
data_fb['platform'] = 'facebook'
data_ig['platform'] = 'instagram'
data_th['platform'] = 'threads'
data_x['platform'] = 'x'
merged_data = pd.concat([data_fb, data_ig, data_th, data_x], axis=0, ignore_index=True)
merged_data
merged_data.to_csv('cleaned_data.csv', index=False)
```
